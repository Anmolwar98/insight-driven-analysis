```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ.get('OPENAI_API_KEY'):
    print('OpenAI API key is set.')

llm_model=ChatOpenAI(model ="gpt-4-0613")
llm_model.invoke('What is the capital of France?')
```

#### **Defined Pydantic Schema**

```python
from pydantic import BaseModel, Field   
class graph_schema(BaseModel):
    topic:str = Field(description = 'This is the LLM Topic')
    post: str = Field(description = 'This is the Linkln post.')
    curated_post: str = Field(description = 'This is the curated post on the same topic')
```

#### **Create Node Function**

```python
def create_post(state: graph_schema) -> graph_schema:
    state = state.model_dump()
    topic = state['topic']
    post = llm_model.invoke(f'Write a Linkedin post on the given topic {topic}').content
    state['post'] = post

    return state
```

```python
def curate_post(state:graph_schema)-> graph_schema:
    state = state.model_dump()
    post = state['post']
    curated_post = llm_model.invoke(f'Curate the following post in a Genz tone: {post}').content
    state['curated_post'] = curated_post

    return state
```

#### **Define StateGraph**

```python
from langgraph.graph import StateGraph, START, END
pydentic_graph = StateGraph(graph_schema)

## add nodes
pydentic_graph.add_node('create_post', create_post)
pydentic_graph.add_node('curate_post', curate_post)

# add edges
pydentic_graph.add_edge(START, 'create_post')
pydentic_graph.add_edge('create_post', 'curate_post')
pydentic_graph.add_edge('curate_post', END)

final_graph = pydentic_graph.compile()
```

#### **Display Graph**

```python
from IPython.display import display, Image
Image(final_graph.get_graph().draw_mermaid_png())
```
