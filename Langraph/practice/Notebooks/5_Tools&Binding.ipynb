##### **Tools & Binding**

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv()

if os.environ.get('OPENAI_API_KEY'):
    print("OPENAI_API_KEY is set.")
else:
    raise ValueError("OPENAI_API_KEY is not set. Please set it in your environment variables.")

llm_model = ChatOpenAI(model='gpt-4-0613')
llm_model.invoke('what is the capital of France?')
```

#### **Create Tools**

```python
from langchain_community.tools import DuckDuckGoSearchRun, WikipediaQueryRun, ArxivQueryRun
from langchain_community.utilities import WikipediaAPIWrapper, ArxivAPIWrapper
from pprint import pprint
from langchain.tools import tool

@tool
def duckduck_search(query:str)->str:
    '''This tool will search the query on duckduckgo and return the result'''
    search = DuckDuckGoSearchRun()
    return search.invoke(query)

@tool
def wikipedia_search(query:str) -> str:
    ''' This tool will search the query on wikipedai and give the detail results'''
    wiki = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())
    return wiki.invoke(query)

@tool
def arxiv_search(query:str) -> str:
    '''This tool will give the latest possible news on the topic'''
    arxiv = ArxivQueryRun(api_wrapper=ArxivAPIWrapper())
    return arxiv.invoke(query)
```

```python
tools = [duckduck_search, wikipedia_search, arxiv_search]

llm_with_tools = llm_model.bind_tools(tools)

response = llm_with_tools.invoke('give the latest news on AI')
pprint(dict(response))
response.tool_calls
```
