#### **Orchestrator**

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv()

if os.environ.get('OPENAI_API_KEY'):
    print('key is set')
else:
    raise ValueError('Key is not set in .env file')

llm_model = ChatOpenAI(model='gpt-4-0613')
llm_model.invoke('what is the 5 problems india is facing today').content
```

#### **LLM Schema**

```python
from typing import List, TypedDict
from pydantic import Field

class llm_schema(TypedDict):
    task: List[str]=Field(description='This create a chunks of task from the given query.')

llm_with_schema = llm_model.with_structured_output(llm_schema)
```

#### **Graph Schema**

```python
from typing import TypedDict, List

class graph_schema(TypedDict):
    query: str
    task: List[str]
    response: List[str]
    summery: str
```

#### **Orchestrator Workflow**

```python
from langchain_core.prompts import ChatPromptTemplate

def orchestrator_node(state:graph_schema):
    user_query = state['query']
    print('user_query-->',user_query)
    prompt = ChatPromptTemplate.from_messages(
        [
            ('system','Youa are an Orchestrator, who will divide the Bigger task into smaller tasks'),
            ('human',f'{user_query}')
        ]
    )
    chain = prompt | llm_with_schema
    response = chain.invoke({'user_query':user_query})
    print('respnse-->',response)
    state['task'] = response['task']
    return state
```

#### **Worker Nodes**

```python
from concurrent.futures import ThreadPoolExecutor

def execute(query):
    return llm_model.invoke(query).content

def worker_node(state:graph_schema):
    tasks = state['task']
    response = []

    with ThreadPoolExecutor(max_workers=len(tasks)) as executor:
        response_results = executor.map(execute, tasks)
        for result in response_results:
            response.append(result)

    state['response'] = response
    return state
```

#### **Collector Node**

```python
def collector_node(state:graph_schema):
    response = state['response']
    prompt = ChatPromptTemplate.from_messages(
        [
            ('system','You are a collector node who get multiple response, based on the responses you get, you need to Summerized it'),
            ('human',f'{response}')
        ]
    )
    chain = prompt | llm_model
    summery = chain.invoke({"response":response}).content
    state['summery']=summery
    return state
```

#### **State Graph**

```python
from langgraph.graph import StateGraph, START, END

graph = StateGraph(graph_schema)

## Add node
graph.add_node('orchestrator_node', orchestrator_node)
graph.add_node('worker_node', worker_node)
graph.add_node('collector_node', collector_node)

## Add edge
graph.add_edge(START, "orchestrator_node")
graph.add_edge('orchestrator_node', "worker_node")
graph.add_edge("worker_node", "collector_node")
graph.add_edge("collector_node", END)

orchestrator_graph = graph.compile()

from IPython.display import Image
Image(orchestrator_graph.get_graph().draw_mermaid_png())
```

```python
orchestrator_graph.invoke({
    'query': 'What is the capital of India, what is the population of Maharashtra and what is the capital of Namibia and what is the population of south Korea',
    'response':[],
    'summery':'',
    'task':[]
})
```
