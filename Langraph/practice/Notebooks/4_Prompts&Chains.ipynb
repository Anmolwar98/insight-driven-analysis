```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv()

if os.environ.get('OPENAI_API_KEY'):
    print("OPENAI_API_KEY is set.")
else:
    raise ValueError("OPENAI_API_KEY is not set. Please set it in your environment variables.")

llm_model = ChatOpenAI(model='gpt-4-0613')
llm_model.invoke('what is the capital of France?')
```

#### **Prompts and Chains**

```python
from langchain_core.prompts import ChatPromptTemplate

my_prompt = ChatPromptTemplate.from_messages(
    [
        ('system','You are a helpful assistat that provide answers professionally.'),
        ('human', 'how to give training to a {topic}.')
    ]
)
```

```python
chain = my_prompt | llm_model
chain.invoke({'topic':'cat'})
```
